```python
#include <iostream>
using namespace std;

int main() {
    string correctUsername = "Preethi";
    string correctPassword = "1234";

    string username, password;
    long long accountNumber;
    int choice;
    float balance = 0, amount;
    cout << "------ LOGIN ------" << endl;
    cout << "Enter Username: ";
    cin >> username;

    cout << "Enter Password: ";
    cin >> password;

    if(username == correctUsername && password == correctPassword) {

        cout << "\nLogin Successful!" << endl;

        cout << "Enter Account Number: ";
        cin >> accountNumber;
        cout << "\n------ BANK MENU ------" << endl;
        cout << "1. Deposit Money" << endl;
        cout << "2. Withdraw Money" << endl;
        cout << "3. Check Balance" << endl;
        cout << "Enter your choice: ";
        cin >> choice;

        switch(choice) {

            case 1:
                cout << "Enter amount to deposit: ";
                cin >> amount;
                balance = balance + amount;
                cout << "Amount Deposited Successfully!" << endl;
                cout << "Current Balance: " << balance << endl;
                break;

            case 2:
                cout << "Enter amount to withdraw: ";
                cin >> amount;

                if(amount <= balance) {
                    balance = balance - amount;
                    cout << "Amount Withdrawn Successfully!" << endl;
                    cout << "Current Balance: " << balance << endl;
                }
                else {
                    cout << "Insufficient Balance!" << endl;
                }
                break;

            case 3:
                cout << "Your Current Balance is: " << balance << endl;
                break;

            default:
                cout << "Invalid Choice!" << endl;
        }

    }
    else {
        cout << "\nInvalid Username or Password!" << endl;
    }

}
```

```python
import math

class Shape:
    def get_area(self):
        return 0

    def display(self):
        print("Generic Shape")


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def get_area(self):
        return self.length * self.width

    def display(self):
        print(f"Rectangle Area = {self.get_area()}")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def get_area(self):
        return math.pi * self.radius ** 2

    def display(self):
        print(f"Circle Area = {self.get_area()}")


def main():
    rect = Rectangle(5, 4)
    circ = Circle(3)

    shapes = [rect, circ]

    for shape in shapes:
        shape.display()


if __name__ == "__main__":
    main()
```

```python
class Member:
    def __init__(self, name, membershipID):
        self.name = name
        self.membershipID = membershipID

    def display(self):
        print(f"Member Type: General")
        print(f"Name: {self.name}")
        print(f"Membership ID: {self.membershipID}")

class RegularMember(Member):
    def __init__(self, name, membershipID):
        super().__init__(name, membershipID)
        self.borrow_limit = 3

    def getBorrowLimit(self):
        return self.borrow_limit

    def display(self):
        super().display()
        print(f"Member Type: Regular")
        print(f"Borrowing Limit: {self.borrow_limit} books")

class PremiumMember(Member):
    def __init__(self, name, membershipID):
        super().__init__(name, membershipID)
        self.borrow_limit = 10
        self.discount = 10

    def getBorrowLimit(self):
        return self.borrow_limit

    def display(self):
        super().display()
        print(f"Member Type: Premium")
        print(f"Borrowing Limit: {self.borrow_limit} books")
        print(f"Discount on membership fees: {self.discount}%")
def main():
    regular_member = RegularMember("John Doe", "R123")
    regular_member.display()
    print(f"Borrow Limit: {regular_member.getBorrowLimit()}")
    premium_member = PremiumMember("Jane Smith", "P456")
    premium_member.display()
    print(f"Borrow Limit: {premium_member.getBorrowLimit()}")

if __name__ == "__main__":
    main()
```

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_number)

class UndergraduateStudent(Student):
    def __init__(self, name, roll_number, m1, m2, m3):
        super().__init__(name, roll_number)
        self.marks = [m1, m2, m3]

    def calculate_average(self):
        return sum(self.marks) / 3

    def display(self):
        print("Undergraduate Student")
        print("Name:", self.name)
        print("Average Marks:", self.calculate_average())

class PostgraduateStudent(Student):
    def __init__(self, name, roll_number, m1, m2, m3, m4):
        super().__init__(name, roll_number)
        self.marks = [m1, m2, m3, m4]

    def calculate_average(self):
        return sum(self.marks) / 4

    def display(self):
        print("Postgraduate Student")
        print("Name:", self.name)
        print("Average Marks:", self.calculate_average())

ug = UndergraduateStudent("Preethi", 101, 85, 90, 88)
pg = PostgraduateStudent("Anu", 201, 80, 75, 85, 90)

ug.display()
print()
pg.display()
```
